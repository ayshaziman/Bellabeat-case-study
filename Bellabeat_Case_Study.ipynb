## Phase 4: Analyze

### Summary Statistics

**Daily Activity:** The dataset contains 457 rows. Users recorded an average of approximately 6,547 steps, 4.66 distance units, and 2,189 calories per day.

**Sleep:** After removing 3 duplicate rows, the dataset contains 410 rows. Users recorded an average of approximately 419 minutes asleep (about 7 hours) and 458 minutes in bed (about 7 hours and 38 minutes).


```python
daily_activity_clean.groupby('ActivityDate')['TotalSteps'].mean()
```

```python
sleep_day_clean['TotalHoursAsleep'] = sleep_day_clean['TotalMinutesAsleep'] / 60

sleep_day_clean[['TotalMinutesAsleep', 'TotalHoursAsleep']].head()
```

```python
sleep_day_clean['TotalHoursAsleep'].mean()
```

### Initial Findings

* Users recorded an average of approximately 6,547 steps per day.
* The highest average daily steps were recorded on April 6, 2016, at approximately 7,989 steps.
* Users slept for an average of approximately 6.99 hours per day.
* Users spent an average of approximately 7 hours and 38 minutes in bed per day.
* Time asleep and time in bed showed a strong positive correlation of approximately 0.93.
* Daily steps and calories burned showed a moderate positive correlation of approximately 0.58.
* Wednesday had the highest average steps and recorded calories, while Tuesday had the lowest averages for both measures.

These are initial observations from the available Fitbit data. Further analysis is needed before making marketing recommendations.



```python
sleep_day_clean['TotalMinutesAsleep'].corr(
    sleep_day_clean['TotalTimeInBed']
)
```

```python
daily_activity_clean['TotalSteps'].corr(
    daily_activity_clean['Calories']
)
```

```python
daily_activity_clean['DayOfWeek'] = daily_activity_clean['ActivityDate'].dt.day_name()

daily_activity_clean.groupby('DayOfWeek')[['TotalSteps', 'Calories']].mean().round(2)
```

```python
daily_activity_clean[['Id', 'ActivityDate', 'TotalSteps']].merge(
    sleep_day_clean[['Id', 'SleepDay', 'TotalMinutesAsleep']],
    left_on=['Id', 'ActivityDate'],
    right_on=['Id', 'SleepDay'],
    how='inner'
).head()
```

```python
daily_activity_clean.to_csv('daily_activity_clean.csv', index=False)
```

```python
sleep_day_clean.to_csv('sleep_day_clean.csv', index=False)
```
